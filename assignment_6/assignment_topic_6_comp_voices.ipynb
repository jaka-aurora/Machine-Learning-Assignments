# Assignment 6.1
## Data preprocessing (2 points)

Familiarize yourself with the data and
implement data preprocessing.

```python
# TODO: Preprocessing data

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('data/voice.csv')

df.dropna(inplace=True)

scaler = StandardScaler()
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
X_scaled = scaler.fit_transform(X)

pd.DataFrame(X_scaled, columns=df.columns[:-1]).head()
```

# Assignment 6.2
## Training and test data (2 points)

Let's split the data into training and test sets and scale the numerical data.

```python
# TODO: Splitting data

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=1550)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

### Decision Tree

```python
# TODO: Decision Tree classification implementation

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix

dt = DecisionTreeClassifier(random_state=1550)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

cr_dt = classification_report(y_test, y_pred_dt)
cm_dt = confusion_matrix(y_test, y_pred_dt)

print(f'Classification report for Decision Tree:\n {cr_dt}')
print(f'Confusion matrix for Decision Tree:\n {cm_dt}')
```

```python
# TODO: Feature importances in Decision Tree classification

feature_importances_dt = dt.feature_importances_
print(f'Feature importances in Decision Tree:\n {feature_importances_dt}')
```

### Support Vector Machine

```python
# TODO: SVC classification implementation

from sklearn.svm import SVC

svm = SVC(kernel='linear', random_state=1550)
svm.fit(X_train, y_train)
y_pred_svm = svm.predict(X_test)

cr_svm = classification_report(y_test, y_pred_svm)
cm_svm = confusion_matrix(y_test, y_pred_svm)

print(f'Classification report for SVM:\n {cr_svm}')
print(f'Confusion matrix for SVM:\n {cm_svm}')
```

### kNN

```python
# TODO: kNN classification implementation

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)

cr_knn = classification_report(y_test, y_pred_knn)
cm_knn = confusion_matrix(y_test, y_pred_knn)

print(f'Classification report for kNN:\n {cr_knn}')
print(f'Confusion matrix for kNN:\n {cm_knn}')
```

### Naive Bayes

```python
# TODO: Naive Bayes classification implementation

from sklearn.naive_bayes import GaussianNB

gnb = GaussianNB()
gnb.fit(X_train, y_train)
y_pred_gnb = gnb.predict(X_test)

cr_gnb = classification_report(y_test, y_pred_gnb)
cm_gnb = confusion_matrix(y_test, y_pred_gnb)

print(f'Classification report for Gaussian Naive Bayes:\n {cr_gnb}')
print(f'Confusion matrix for Gaussian Naive Bayes:\n {cm_gnb}')
```

### Random Forest

```python
# TODO: Random Forest classification implementation

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=1550)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

# Classification report and confusion matrix for Random Forest
cr_rf = classification_report(y_test, y_pred_rf)
cm_rf = confusion_matrix(y_test, y_pred_rf)

print(f'Classification report for Random Forest:\n {cr_rf}')
print(f'Confusion matrix for Random Forest:\n {cm_rf}')

# Feature importances in Random Forest classification
feature_importances_rf = rf.feature_importances_
print(f'Feature importances in Random Forest:\n {feature_importances_rf}')
```

### PCA + classifier
PCA with *Random Forest Classifier* or any other classifier you want to use.

```python
# TODO: PCA + selected Classifier - using Random Forest

from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(X_pca, y, test_size=0.2, random_state=1550)

rf_pca = RandomForestClassifier(random_state=1550)
rf_pca.fit(X_train_pca, y_train_pca)
y_pred_rf_pca = rf_pca.predict(X_test_pca)

cr_rf_pca = classification_report(y_test_pca, y_pred_rf_pca)
cm_rf_pca = confusion_matrix(y_test_pca, y_pred_rf_pca)

print(f'Classification report for PCA + Random Forest:\n {cr_rf_pca}')
print(f'Confusion matrix for PCA + Random Forest:\n {cm_rf_pca}')
```
